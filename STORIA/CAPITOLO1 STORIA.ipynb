***IL MONDO E L'ITALIA NEL SECONDO OTTOCENTO***

****RECAP****

-*1914-1918*

*DUE BLOCCHI TRIPLICE ALLENZA E TRIPLICE IMPRESA*

-prima alleanza:germani austria e italia

-seconda alleanza:inghilterra francia e russia

**BELLEPOCQUE 1875-1910**

*chiamata anche età d'oro dopo la prima guerra mondiale*

**PERIODO DI PROGREESSO**

*conquiste in innovazioni tecnologiche e vaccini scoperta dell'uranio e dell'acciaio*

**SI SCOPRE IL TEMPO LIBERO**

*ciclismo,e sport durante il tempo libero ma anche la psicoanalisi*

**DURANTE LA SECONDA RIVOLUZIONE INDUSTRIALE**

*si potenzia la moneta con le banche coniando o stampando, cambiando il valore della moneta alla quantità di oro presenti nelle riserve delle banche nazionali direttamente proporzionale*

**AUMENTA L'INDUSTRALIZZAZIONE**

*ai proprietari del'industria conveniva la guerra per guadagno dopo lo sgancio della bomba ne serviva un'altra**

**IL COLONIALISMO**

*DIFFERENZA TRA QUELLO DEL 500 E QUELLO DELL'OTTOCENTO*

*tra il primo colonialismo e il secondo il primo colonialismno si concentrava sul prelievo di risorse del territorio, gli stati coloniali del secondo devono occupare le terre per risorse specifiche* dove si stabiliscono accordi tra le persone presenti in quel territorio* 

*nella seconda metà dell'ottocento ci sono le colonie gia formate e quelle che si devono già formare tra quele gia formate abbiamo:francia e inghilterra per esempio e in quelle che si ovranno formare ci sono colonie come italia o la prussia che diventerà germania*

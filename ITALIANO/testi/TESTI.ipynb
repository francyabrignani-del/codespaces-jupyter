***IL NATURALISMO***

Il Naturalismo nasce in Francia negli anni Settanta dell’Ottocento e sarà
  importante anche per la nascita del Verismo italiano.

  🔬 I fondamenti teorici

  Alla base del Naturalismo c’è il Positivismo, corrente di pensiero legata
  allo sviluppo della società industriale, della scienza e della tecnologia.
  Secondo il Positivismo, la realtà deve essere studiata attraverso un metodo
  scientifico, cercando leggi precise che spieghino i fenomeni.

  Un pensatore fondamentale fu Hippolyte Taine, sostenitore del determinismo
  materialistico. Secondo lui, il comportamento e la personalità dell’uomo
  sono determinati soprattutto da tre elementi:

  razza + ambiente + momento storico

  Per questo anche la letteratura deve diventare una sorta di indagine
  scientifica sull’uomo e sulla società: lo scrittore osserva e analizza la
  realtà con precisione, quasi come uno scienziato.

  ⸻

  ✍️ I precursori del Naturalismo

  Prima della piena affermazione del Naturalismo, alcuni scrittori ne
  anticiparono diversi principi.

  Honoré de Balzac viene considerato da Taine un modello di “scrittore
  scienziato”, perché nelle sue opere analizza attentamente la società e la
  natura umana.

  Gustave Flaubert, autore di Madame Bovary (1857), sviluppa invece la teoria
  dell’impersonalità: lo scrittore deve essere presente nell’opera senza
  intervenire direttamente con giudizi e sentimenti personali. Deve quindi
  rappresentare la realtà nel modo più oggettivo e preciso possibile.

  Infine, i fratelli Edmond e Jules de Goncourt basano i loro romanzi su una
  documentazione minuziosa e diretta della società. Si interessano soprattutto
  alle classi inferiori, alla degradazione umana e ai casi patologici. Nel
  romanzo Germinie Lacerteux (1865) sostengono inoltre che anche le classi
  popolari abbiano il diritto di essere rappresentate nella letteratura.
